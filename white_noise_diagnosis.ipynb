```python
importlib.reload(cf)
importlib.reload(FP)
filepath = Path(r"D:\Goffaux lab\Data set\second_expt\TRSG1512.npy")
savefilepath = Path(r"D:\Goffaux lab\Data set\second_expt\white_noise.npy")

npy_data,meta_data, _, _ = FP.loadalldata(filepath)

fs = meta_data['fs']            # sampling rate (Hz)
duration = npy_data.shape[0]//fs      # seconds
nch = npy_data.shape[1]            # channels

sigma = 10.0
n = fs * duration                                    # number of time samples
f = np.fft.rfftfreq(n, 1 / fs)                       # frequency axis: 0 ... fs/2 Hz
rng = np.random.default_rng(seed=1)

A = sigma * np.sqrt(n)                               # flat magnitude that gives std = sigma in time
spectrum = A * np.exp(1j * rng.uniform(0, 2 * np.pi, (len(f), nch)))   # flat magnitude, random phase
spectrum[0] = 0                                      # DC = 0 → zero-mean signal
if n % 2 == 0:
    spectrum[-1] = A                                 # Nyquist bin must be real

spectrum = spectrum*(1/.0075)
npy_data_white = np.fft.irfft(spectrum, n=n, axis=0)     # optional: back to time, shape (samples, channels)

FP.savealldata(npy_data_white,meta_data,savefilepath)
```
